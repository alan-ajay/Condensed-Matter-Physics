# **Free electron band structure**

Recall the derivation of $H_k u_k = E_k u_k$ in class.
Consider switching the amplitude of the periodic potential slowly all the way to zero with the notion of periodicity "a" still intact.

---



 Q1. Assume $u_m = exp(\frac{i m 2 \pi x}{a})$ in 1D, plot $E_k$ for $k = -6 π/a$ to $+6 π/a$. Correlate m with band index.

 Q2.    Plot $E_k$ for the same interval for periodicity "2a" and compare with your plot in Q1


```python
import numpy as np
import matplotlib.pyplot as plt

a = 2.0
k = np.linspace(-6 * np.pi / a, 6 * np.pi / a, 1000)
m_values = np.arange(-6, 7)
m_values_2a = np.arange(-9, 10)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

for m in m_values:
    E_a = (k + (2 * np.pi * m) / a)**2
    ax1.plot(k, E_a, lw=1.5)

bz_limit = np.pi / a
energies = [(np.pi/a)**2, (2*np.pi/a)**2, (3*np.pi/a)**2, (4*np.pi/a)**2]
ax1.hlines(y=0, xmin=-bz_limit, xmax=bz_limit, colors='red', linestyles='--', label='Band Index')
for i, e in enumerate(energies):
    ax1.hlines(y=e, xmin=-bz_limit, xmax=bz_limit, colors='red', linestyles='--')
    ax1.text(0, e + 0.5, f"n={i+1}", color='red', ha='center', fontweight='bold')

ax1.set_title(r"Periodicity $a$")
ax1.set_xlabel(r"k")
ax1.set_ylabel(r"$E_k$")
ax1.set_ylim(0, 50)
ax1.axvline(x=-bz_limit, color='black', linestyle='--', label=r'1st BZ boundary($\pm \pi/a$)')
ax1.axvline(x=bz_limit, color='black', linestyle='--')
ax1.grid(True, alpha=0.3)
ax1.legend()

for m in m_values_2a:
    E_2a = (k + (2 * np.pi * m) / (2 * a))**2
    ax2.plot(k, E_2a, lw=1.5)

ax2.set_title(r"Periodicity $2a$")
ax2.set_xlabel(r"k")
ax2.set_ylim(0, 50)
ax2.axvline(x=-np.pi/(2*a), color='black', linestyle='--', label='New 1st BZ boundary($\pm \pi/2a$)')
ax2.axvline(x=np.pi/(2*a), color='black', linestyle='--')
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.suptitle(f"Band structure for free electrons: a={a}", fontsize=16)
plt.tight_layout()
plt.show()


```



## 1. Correlation of m vs n for 1st BZ
